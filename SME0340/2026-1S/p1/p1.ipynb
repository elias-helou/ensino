<h1 style="margin-bottom: 0;">Questão 1</h1>
<strong>Valor 1.0</strong>

Considere a equação diferencial ordinária dada por
$$
    \frac{\mathrm dy}{\mathrm dx} = e^{-x^2}.
$$
Sem resolver a EDO, responda os itens abaixo:
1. Explique por que a solução da ED necessariamente é uma função crescente em qualquer intervalo do eixo $x$.

2. Quais são os limites $\displaystyle\lim_{x\to-\infty}dy/dx$ e $\displaystyle\lim_{x\to\infty}dy/dx$? O que isto implica a respeito da curva da solução quando $x\to\pm\infty$?

3. Determine um intervalo no qual a solução $y=\phi(x)$ tem concavidade para baixo e um intervalo no qual a curva tem concavidade para cima.

## Item 1

A solução da EDO é uma função $y( x )$ tal que $y'( x ) = e^{-x^2} > 0$. Ou seja, a derivada da solução é sempre positiva, portanto, trata-se de uma função estritamente crescente em toda a reta real.

## Item 2

A mera inspeção da EDO nos garante que
$$
    \lim_{x\to\infty}\frac{\mathrm dy}{\mathrm dx} = 0\quad \text e\quad \lim_{x\to-\infty}\frac{\mathrm dy}{\mathrm dx} = 0.
$$
Isto implica que qualquer solução da EDO possui uma assíntota horizontal à esquerda e outra à direita.

## Item 3

A concavidade do gráfico de uma função depende de sua segunda derivada. Diferenciação de ambos os lados da EDO nos dá
$$
    \frac{\mathrm dy^2}{\mathrm d^2x} = -2xe^{-x^2}.
$$

Portanto, a curva do gráfico de $y$ possui concavidade para cima em $x \in ( -\infty, 0 )$ e para baixo em $x \in ( 0, \infty )$. O ponto $x = 0$ é um ponto de inflexão.

## Item 1

Observe que
$$
    y' = \frac 1{( x + c )^2}
$$
e que
$$
    y^2 = \frac 1{( x + c )^2}.
$$
Portanto,
$$
 y' = y^2.
$$

## Item 2


Veja que, se $y( 0 ) = 1$, então
$$
    \frac {-1}{0 + c} = 1 \Leftrightarrow c = -1.
$$
Ou seja,
$$
    y( x ) = \frac {-1}{x - 1}.
$$
O intervalo de definição neste caso é, portanto,
$$
    I = (-\infty, 1 ).
$$

Agora, se $y( 0 ) = -1$, então
$$
    \frac {-1}{0 + c} = -1 \Leftrightarrow c = 1.
$$
Ou seja,
$$
    y( x ) = \frac {-1}{x + 1}.
$$
O intervalo de definição neste caso é, portanto,
$$
    I = ( -1, \infty ).
$$


## Item 3

Veja que não existe constante para a qual
$$
    \frac {-1}{0 + c} = 0.
$$
Entretanto, uma solução trivial para a EDO que satisfaz as condições iniciais dada é
$$
    y( x ) \equiv 0,
$$
cujo intervalo de definição é:
$$
    I = \mathbb R.
$$

```python
x, y, c = sp.symbols( 'x, y, c' )
```

```python
def separavel( f, g, checa_sol = False ):

    ipd.display( ipd.Markdown( 'Nossa equação tem a forma' ) )
    ipd.display(
        ipd.Math(
            sp.latex( f ) + r'\frac{\mathrm dy}{\mathrm dx} = ' + sp.latex( g ) + '.'
        )
    )
    if ( sp.diff( f, x ) == 0 ) and ( sp.diff( g, y ) == 0 ):
        ipd.display( ipd.Markdown( 'Portanto, é uma EDO separável.' ) )
    else:
        ipd.display( ipd.Markdown( 'Portanto, não é uma EDO separável. Abortando!' ) )
        return

    F = sp.integrate( f, y )
    G = sp.integrate( g, x )

    ltx_str = ''

    ltx = r'\int' + sp.latex( f ) + r'\mathrm dy = ' + sp.latex( F )
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )
    ltx = r'\int' + sp.latex( g ) + r'\mathrm dx = ' + sp.latex( G )
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )

    imp_sol = sp.Eq( F, G + c )
    ltx = sp.latex( imp_sol )
    ltx_str += ltx + '\n'

    ipd.display(
        ipd.Markdown(
            r'A solução implícita é, portanto,$$' + ltx + '.$$'
        )
    )

    ipd.display( ipd.Markdown( r'Resolvendo para $y$, obtemos:' ) )
    sols = sp.solve( imp_sol, y )
    for s in sols:
        ltx = 'y( x ) = ' + sp.latex( s )
        ltx_str += ltx + '\n'
        ipd.display( ipd.Math( ltx ) )

    # Verificação das soluções:
    if checa_sol:
        ipd.display(
            ipd.Markdown(
                r'Calculando $$' + sp.latex( f ) + r'\frac{\mathrm dy}{\mathrm dx} - ' + sp.latex( g ) +  r',$$obtemos:'
            )
        )
        for s in sols:
            chk =  sp.simplify(
                        f.subs( y, s ) * sp.diff( s, x ) - g.subs( y, s )
                    )
            ltx = sp.latex( chk )
            ltx_str += ltx + '\n'
            ipd.display(
                ipd.Math(
                    ltx
                )
            )

    return ltx_str, imp_sol, sols
```

```python
def exata( M, N, checa_sol = False ):

    ipd.display( ipd.Markdown( 'Nossa equação tem a forma' ) )
    ipd.display(
        ipd.Math(
            r'M( x, y ) + N( x, y ) \frac{\mathrm dy}{\mathrm dx} = ' + \
            sp.latex( M ) + r'+\left (' + sp.latex( N ) + r'\right )\frac{\mathrm dy}{\mathrm dx} = 0.'
        )
    )
    M_y = sp.diff( M, y )
    N_x = sp.diff( N, x )

    ltx_str = ''

    ltx = r'M_y = ' + sp.latex( M_y )
    ltx_str += ltx + r'\quad \text{e} \quad '

    ltx = r'N_x = ' + sp.latex( N_x )
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx_str ) )

    if ( sp.simplify( M_y - N_x ) ) != 0:
        ipd.display( ipd.Markdown( r'Como $M_y \neq N_x$, não temos uma equação exata. Abortando execução!' ) )
        return

    ipd.display( ipd.Markdown( r'Como $M_y = N_x$, temos uma equação exata. Assim, podemos encontrar uma solução implícita por integração.' ) )

    G_part = sp.integrate( M, x )
    ltx = r'\int M\mathrm dx = ' + sp.latex( G_part ) + r' + g( y )'
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )

    g_prime = N - sp.diff( G_part, y )
    g = sp.integrate( g_prime, y )
    ltx = r'g( y ) = \int \left( N - \frac{\mathrm d}{\mathrm dy}\int M\mathrm dx \right) \mathrm d y = \int ' + sp.latex( g_prime ) + r'\mathrm dy = ' + sp.latex( g ) + ' - c'
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )

    G = G_part + g
    imp_sol = sp.Eq( G, c )
    ltx = sp.latex( imp_sol )
    ltx_str += ltx + '\n'

    ipd.display(
        ipd.Markdown(
            r'A solução implícita é, portanto,$$' + ltx + '.$$'
        )
    )

    ipd.display( ipd.Markdown( r'Resolvendo para $y$, obtemos:' ) )
    sols = sp.solve( imp_sol, y )
    for s in sols:
        ltx = 'y( x ) = ' + sp.latex( s )
        ltx_str += ltx + '\n'
        ipd.display( ipd.Math( ltx ) )

    # Verificação das soluções:
    if checa_sol:
        ipd.display(
            ipd.Markdown(
                r'Calculando $$' + sp.latex( M ) + r'+\left (' + sp.latex( N ) + r'\right )\frac{\mathrm dy}{\mathrm dx},' +  r'$$obtemos:'
            )
        )
        for s in sols:
            chk =  sp.simplify(
                        M.subs( y, s ) + N.subs( y, s ) * sp.diff( s, x )
                    )
            ltx = sp.latex( chk )
            ltx_str += ltx + '\n'
            ipd.display(
                ipd.Math(
                    ltx
                )
            )

    return ltx_str, imp_sol, sols
```

2. Mostre que as condições iniciais $y(0)=-2$ e $y(1)=1$ determinam a mesma solução implícita.

## Item 2

Note que $y( 0 ) = -2$, na solução implícita, resulta em
$$
    0^3 + 2\cdot0^2\cdot( -2 ) + ( -2 )^2 = c \Leftrightarrow c = 4.
$$

Por outro lado, $y( 1 ) = 1$, na solução implícita, resulta em
$$
    1^3 + 2\cdot1^2\cdot 1 + 1^2 = c \Leftrightarrow c = 4.
$$

## Item 3

A solução explícita pode ser calculada como a raiz de uma equação de segundo grau em $y$:
$$
    y = \frac{-2x^2 \pm \sqrt{4x^4 - 4(x^3 - c)}}{2} = -x^2 \pm \sqrt{x^4 - x^3 + c}.
$$

Ao utilizarmos a forma explícita para encontrar a solução que satisfaz $y( 0 ) = -2$, obtemos
$$
    \pm\sqrt{c} = -2 \Leftrightarrow c = 4.
$$
Desta vez, está claro que precisamos usar o ramo negativo da raiz quadrada, ou seja,
$$
    y( x ) = -x^2 - \sqrt{x^4 + 4}.
$$

Por outro lado, $y( 1 ) = 1$, na solução explícita, resulta em
$$
    -1 \pm \sqrt{1 - 1 + c} = 1 \Leftrightarrow c = 4.
$$
Desta vez, está claro que precisamos usar o ramo positivo da raiz quadrada, ou seja,
$$
    y( x ) = -x^2 + \sqrt{x^4 + 4}.
$$


```python
def linear( P, f, checa_sol = False ):

    ipd.display( ipd.Markdown( 'Nossa equação tem a forma' ) )
    ipd.display(
        ipd.Math(
            r'\frac{\mathrm dy}{\mathrm dx} + \left(' + sp.latex( P ) +  r'\right)y = ' + sp.latex( f )
        )
    )

    G = sp.exp( sp.integrate( P, x ) )

    ltx_str = ''

    ltx = r'G( x ) = ' + sp.latex( G )
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx_str ) )

    s = ( sp.integrate( f * G, x ) + c ) / G
    ltx = r'\begin{aligned} y( x ) &{}= \frac{\int G( x )f( x ) \mathrm dx}{G( x )}\\'
    ltx += r'& {}= \frac{\int ' + sp.latex( G * f ) + r'\mathrm dx}{' + sp.latex( G ) + r'}\\'
    ltx += r'& {}= ' + sp.latex( s )
    ltx += r'\end{aligned}'
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )

    if checa_sol:
        ltx = sp.latex(
            sp.simplify(
                sp.diff( s, x ) + P * s - f
            )
        )
        ipd.display( ipd.Math( ltx ) )

    return ltx_str
```

Um marca-passo cardíaco consiste em uma chave, uma bateria de tensão constante $E_0$, um capacitor com capacitância constante $C$, e o coração, agindo como um resistor com resistência constante $R$. Quando a chave é fechada, o capacitor se carrega; quando a chave é aberta, o capacitor se descarrega, enviando um estímulo elétrico para o coração. Durante o tempo em que  o coração é estimulado, a tensão $E$ em todo o coração satisfaz a equação diferencial linear
$$
    \frac{dE}{dt}=-\frac{1}{RC}E
$$
Resolva a EDO acima e o PVI associado com condição inicial $E(4)=E_0$.


```python
R, C = sp.symbols( 'R, C' )

linear( 1 / ( R * C ), 0 );
```

Agora lidaremos com a condição inicial $E( 4 ) = E_0$. Isto equivale a
$$
    ce^{-\frac{4}{CR}} = E_0 \Leftrightarrow c = E_0e^{\frac{4}{CR}}.
$$
Portanto, a solução do PVI é
$$
    E( t ) = E_0e^{\frac{4}{CR}}e^{-\frac{t}{CR}} = E_0e^{\frac{4 - t}{CR}}.
$$

```python
def bernoulli( P, f, n, checa_sol = False ):

    ipd.display( ipd.Markdown( 'Nossa equação tem a forma' ) )
    ipd.display(
        ipd.Math(
            r'\frac{\mathrm dy}{\mathrm dx} + \left(' + sp.latex( P ) +  r'\right) y = ' + sp.latex( f * y ** n )
        ) 
    )

    ipd.display(
        ipd.Markdown(
            'Através da subsituição $u = y^{1 - ' + sp.latex( n ) + '}$, obtemos a EDO linear:'
        )
    )
    ber_exp = r'\left( 1 - ' + sp.latex( n ) + r'\right )'
    ipd.display(
        ipd.Math(
            r'\frac{\mathrm du}{\mathrm dx} + ' + ber_exp + r' \left(' + sp.latex( P ) +  r'\right) u = ' + ber_exp + r'\left(' + sp.latex( f ) + r'\right)'
        )
    )

    G = sp.exp( sp.integrate( ( 1 - n ) * P, x ) )

    ltx_str = ''

    ltx = r'G( x ) = ' + sp.latex( G )
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx_str ) )

    u = ( sp.integrate( ( 1 - n ) * f * G, x ) + c ) / G
    ltx = r'\begin{aligned} u( x ) &{}= \frac{\int G( x )' + ber_exp + r'f( x ) \mathrm dx}{G( x )}\\'
    ltx += r'& {}= \frac{\int ' + sp.latex( G * ( 1 - n ) * f ) + r'\mathrm dx}{' + sp.latex( G ) + r'}\\'
    ltx += r'& {}= ' + sp.latex( u )
    ltx += r'\end{aligned}'
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )
    s = u ** ( 1 / ( 1 - n ) )

    if checa_sol:
        ltx = sp.latex(
            sp.simplify(
                sp.diff( u, x ) + ( 1 - n ) * P * u - ( 1 - n ) * f
            )
        )
        ipd.display( ipd.Math( ltx ) )
        ltx = sp.latex(
            sp.simplify(
                sp.diff( s, x ) + P * s - f * ( s ** n )
            )
        )
        ipd.display( ipd.Math( ltx ) )

    return ltx_str, s
```
