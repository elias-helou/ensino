```python
import sympy as sp
import IPython.display as ipd
```

```python
x, y, c = sp.symbols( 'x, y, c' )
```

```python
def separavel( f, g, checa_sol = False ):

    ipd.display( ipd.Markdown( 'Nossa equação tem a forma' ) )
    ipd.display(
        ipd.Math(
            sp.latex( f ) + r'\frac{\mathrm dy}{\mathrm dx} = ' + sp.latex( g ) + '.'
        )
    )
    if ( sp.diff( f, x ) == 0 ) and ( sp.diff( g, y ) == 0 ):
        ipd.display( ipd.Markdown( 'Portanto, é uma EDO separável.' ) )
    else:
        ipd.display( ipd.Markdown( 'Portanto, não é uma EDO separável. Abortando!' ) )
        return

    F = sp.integrate( f, y )
    G = sp.integrate( g, x )

    ltx_str = ''

    ltx = r'\int' + sp.latex( f ) + r'\mathrm dy = ' + sp.latex( F )
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )
    ltx = r'\int' + sp.latex( g ) + r'\mathrm dx = ' + sp.latex( G )
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )

    imp_sol = sp.Eq( F, G + c )
    ltx = sp.latex( imp_sol )
    ltx_str += ltx + '\n'

    ipd.display(
        ipd.Markdown(
            r'A solução implícita é, portanto,$$' + ltx + '.$$'
        )
    )

    ipd.display( ipd.Markdown( r'Resolvendo para $y$, obtemos:' ) )
    sols = sp.solve( imp_sol, y )
    for s in sols:
        ltx = 'y( x ) = ' + sp.latex( s )
        ltx_str += ltx + '\n'
        ipd.display( ipd.Math( ltx ) )

    # Verificação das soluções:
    if checa_sol:
        ipd.display(
            ipd.Markdown(
                r'Calculando $$' + sp.latex( f ) + r'\frac{\mathrm dy}{\mathrm dx} - ' + sp.latex( g ) +  r',$$obtemos:'
            )
        )
        for s in sols:
            chk =  sp.simplify(
                        f.subs( y, s ) * sp.diff( s, x ) - g.subs( y, s )
                    )
            ltx = sp.latex( chk )
            ltx_str += ltx + '\n'
            ipd.display(
                ipd.Math(
                    ltx
                )
            )

    return ltx_str, imp_sol, sols
```

```python
ltx, imp_sol, sols = separavel( y / ( y + 1 ) ** 2, 1 / ( x * sp.ln( x ) ), True )
```

```python
ltx, imp_sol, sols = separavel( ( y - 2 ) / ( y + 3 ), ( x - 1 ) / ( x + 4 ), True )
```

```python
def exata( M, N, checa_sol = False ):

    ipd.display( ipd.Markdown( 'Nossa equação tem a forma' ) )
    ipd.display(
        ipd.Math(
            r'M( x, y ) + N( x, y ) \frac{\mathrm dy}{\mathrm dx} = ' + \
            sp.latex( M ) + r'+\left (' + sp.latex( N ) + r'\right )\frac{\mathrm dy}{\mathrm dx} = 0.'
        )
    )
    M_y = sp.diff( M, y )
    N_x = sp.diff( N, x )

    ltx_str = ''

    ltx = r'M_y = ' + sp.latex( M_y )
    ltx_str += ltx + r'\quad \text{e} \quad '

    ltx = r'N_x = ' + sp.latex( N_x )
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx_str ) )

    if ( sp.simplify( M_y - N_x ) ) != 0:
        ipd.display( ipd.Markdown( r'Como $M_y \neq N_x$, não temos uma equação exata. Abortando execução!' ) )
        return

    ipd.display( ipd.Markdown( r'Como $M_y = N_x$, temos uma equação exata. Assim, podemos encontrar uma solução implícita por integração.' ) )

    G_part = sp.integrate( M, x )
    ltx = r'\int M\mathrm dx = ' + sp.latex( G_part ) + r' + g( y )'
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )

    g_prime = N - sp.diff( G_part, y )
    g = sp.integrate( g_prime, y )
    ltx = r'g( y ) = \int \left( N - \frac{\mathrm d}{\mathrm dy}\int M\mathrm dx \right) \mathrm d y = \int ' + sp.latex( g_prime ) + r'\mathrm dy = ' + sp.latex( g ) + ' - c'
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )

    G = G_part + g
    imp_sol = sp.Eq( G, c )
    ltx = sp.latex( imp_sol )
    ltx_str += ltx + '\n'

    ipd.display(
        ipd.Markdown(
            r'A solução implícita é, portanto,$$' + ltx + '.$$'
        )
    )

    ipd.display( ipd.Markdown( r'Resolvendo para $y$, obtemos:' ) )
    sols = sp.solve( imp_sol, y )
    for s in sols:
        ltx = 'y( x ) = ' + sp.latex( s )
        ltx_str += ltx + '\n'
        ipd.display( ipd.Math( ltx ) )

    # Verificação das soluções:
    if checa_sol:
        ipd.display(
            ipd.Markdown(
                r'Calculando $$' + sp.latex( M ) + r'+\left (' + sp.latex( N ) + r'\right )\frac{\mathrm dy}{\mathrm dx},' +  r'$$obtemos:'
            )
        )
        for s in sols:
            chk =  sp.simplify(
                        M.subs( y, s ) + N.subs( y, s ) * sp.diff( s, x )
                    )
            ltx = sp.latex( chk )
            ltx_str += ltx + '\n'
            ipd.display(
                ipd.Math(
                    ltx
                )
            )

    return ltx_str, imp_sol, sols
```

```python
ltx_str, imp_sol, sols = exata( 4 * x * y + 3 * x ** 2, 2 * y + 2 * x ** 2, True )
```

```python
def linear( P, f, checa_sol = False ):

    ipd.display( ipd.Markdown( 'Nossa equação tem a forma' ) )
    ipd.display(
        ipd.Math(
            r'\frac{\mathrm dy}{\mathrm dx} + \left(' + sp.latex( P ) +  r'\right)y = ' + sp.latex( f )
        )
    )

    G = sp.exp( sp.integrate( P, x ) )

    ltx_str = ''

    ltx = r'G( x ) = ' + sp.latex( G )
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx_str ) )

    s = ( sp.integrate( f * G, x ) + c ) / G
    ltx = r'\begin{aligned} y( x ) &{}= \frac{\int G( x )f( x ) \mathrm dx}{G( x )}\\'
    ltx += r'& {}= \frac{\int ' + sp.latex( G * f ) + r'\mathrm dx}{' + sp.latex( G ) + r'}\\'
    ltx += r'& {}= ' + sp.latex( s )
    ltx += r'\end{aligned}'
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )

    if checa_sol:
        ltx = sp.latex(
            sp.simplify(
                sp.diff( s, x ) + P * s - f
            )
        )
        ipd.display( ipd.Math( ltx ) )

    return ltx_str
```

```python
R, C = sp.symbols( 'R, C' )

linear( 1 / ( R * C ), 0, True );
```

```python
def bernoulli( P, f, n, checa_sol = False ):

    ipd.display( ipd.Markdown( 'Nossa equação tem a forma' ) )
    ipd.display(
        ipd.Math(
            r'\frac{\mathrm dy}{\mathrm dx} + \left(' + sp.latex( P ) +  r'\right) y = ' + sp.latex( f * y ** n )
        ) 
    )

    ipd.display(
        ipd.Markdown(
            'Através da subsituição $u = y^{1 - ' + sp.latex( n ) + '}$, obtemos a EDO linear:'
        )
    )
    ber_exp = r'\left( 1 - ' + sp.latex( n ) + r'\right )'
    ipd.display(
        ipd.Math(
            r'\frac{\mathrm du}{\mathrm dx} + ' + ber_exp + r' \left(' + sp.latex( P ) +  r'\right) u = ' + ber_exp + r'\left(' + sp.latex( f ) + r'\right)'
        )
    )

    G = sp.exp( sp.integrate( ( 1 - n ) * P, x ) )

    ltx_str = ''

    ltx = r'G( x ) = ' + sp.latex( G )
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx_str ) )

    u = ( sp.integrate( ( 1 - n ) * f * G, x ) + c ) / G
    ltx = r'\begin{aligned} u( x ) &{}= \frac{\int G( x )' + ber_exp + r'f( x ) \mathrm dx}{G( x )}\\'
    ltx += r'& {}= \frac{\int ' + sp.latex( G * ( 1 - n ) * f ) + r'\mathrm dx}{' + sp.latex( G ) + r'}\\'
    ltx += r'& {}= ' + sp.latex( u )
    ltx += r'\end{aligned}'
    ltx_str += ltx + '\n'
    ipd.display( ipd.Math( ltx ) )
    s = u ** ( 1 / ( 1 - n ) )

    if checa_sol:
        ltx = sp.latex(
            sp.simplify(
                sp.diff( u, x ) + ( 1 - n ) * P * u - ( 1 - n ) * f
            )
        )
        ipd.display( ipd.Math( ltx ) )
        ltx = sp.latex(
            sp.simplify(
                sp.diff( s, x ) + P * s - f * ( s ** n )
            )
        )
        ipd.display( ipd.Math( ltx ) )

    return ltx_str, s
```

```python
alpha, beta = sp.symbols( r'\alpha, \beta' )
ltx, sol = bernoulli( beta, alpha, sp.Rational( 2, 3 ), True )
```
